````python
import os
import json
import requests
from dotenv import load_dotenv

# Forzar la recarga de las variables desde el archivo .env actualizado
load_dotenv(override=True)
API_KEY = os.getenv("YOUTUBE_API_KEY")

def obtener_datos_canal_youtube(handle="@mouredev"):
    if not API_KEY or "AIzaSy" not in API_KEY or "90VpO" in API_KEY:
        print("⚠️ Advertencia: Parece que sigues usando la clave de prueba o no está configurada.")

    url = "https://www.googleapis.com/youtube/v3/channels"

    params = {
        "part": "snippet,statistics,contentDetails",
        "forHandle": handle,
        "key": API_KEY
    }

    print(f"🔄 Consultando los datos del canal {handle} en YouTube...")

    try:
        response = requests.get(url, params=params, timeout=10)
        response.raise_for_status()

        data = response.json()
        items = data.get("items", [])

        if not items:
            print(f"⚠️ No se encontró ningún canal asociado al handle: {handle}")
            return

        print("✅ ¡Datos del canal obtenidos exitosamente!")

        canal_info = items[0]
        snippet = canal_info.get("snippet", {})
        statistics = canal_info.get("statistics", {})

        datos_procesados = {
            "channel_id": canal_info.get("id"),
            "titulo": snippet.get("title"),
            "descripcion": snippet.get("description"),
            "custom_url": snippet.get("customUrl"),
            "fecha_creacion": snippet.get("publishedAt"),
            "pais": snippet.get("country"),
            "suscriptores": statistics.get("subscriberCount"),
            "vistas_totales": statistics.get("viewCount"),
            "videos_totales": statistics.get("videoCount")
        }

        nombre_archivo = "mouredev_channel_data.json"
        with open(nombre_archivo, "w", encoding="utf-8") as f:
            json.dump(datos_procesados, f, indent=4, ensure_ascii=False)

        print(f"💾 Información guardada correctamente en '{nombre_archivo}'.")

    except requests.exceptions.HTTPError as http_err:
        print(f"❌ Error HTTP: {http_err} (Código de estado: {response.status_code})")
        if response.status_code == 400:
            print("💡 Consejo: Tu clave API sigue siendo inválida. Asegúrate de poner una clave real activa en el archivo .env")
    except Exception as err:
        print(f"❌ Ocurrió un error inesperado: {err}")

if __name__ == "__main__":
    obtener_datos_canal_youtube(handle="@mouredev")
````
